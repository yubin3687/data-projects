```python
import pandas as pd

# 파일 경로는 본인 환경에 맞게 수정
df = pd.read_csv('C:/Users/USER/Desktop/data/inventory_items.csv')

def create_detailed_data_spec(df):
    spec = pd.DataFrame({
        '컬럼명': df.columns,
        '데이터 타입': df.dtypes.values,
        '결측값 개수': df.isnull().sum().values,
        '고유값 개수': df.nunique().values,
        '최소 값': [df[col].min() if pd.api.types.is_numeric_dtype(df[col]) else '' for col in df.columns],
        '최대 값': [df[col].max() if pd.api.types.is_numeric_dtype(df[col]) else '' for col in df.columns],
        '첫 번째 값': [df[col].dropna().iloc[0] if len(df[col].dropna()) > 0 else '' for col in df.columns],
        '두 번째 값': [df[col].dropna().iloc[1] if len(df[col].dropna()) > 1 else '' for col in df.columns],
        '세 번째 값': [df[col].dropna().iloc[2] if len(df[col].dropna()) > 2 else '' for col in df.columns],
        '끝 값': [df[col].dropna().iloc[-1] if len(df[col].dropna()) > 0 else '' for col in df.columns]
    })
    return spec

data_spec = create_detailed_data_spec(df)
data_spec
```
