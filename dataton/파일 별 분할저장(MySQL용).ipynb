### MySQL에 넣기 좋게 10000행씩 분할 저장

```python
# 1. 전처리된 파일 불러오기
df = pd.read_csv('C:/Users/USER/Desktop/data/cleaned/inventory_items_cleaned.csv')

# 2. 한 파일당 행 수 설정
chunk_size = 10000
num_chunks = math.ceil(len(df) / chunk_size)

# 3. 분할 저장
for i in range(num_chunks):
    start = i * chunk_size
    end = start + chunk_size
    chunk = df[start:end]
    
    filename = f'C:/Users/USER/Desktop/data/forMySQL/inventory_items_cleaned{i+1}.csv'
    chunk.to_csv(filename, index=False)
    print(f'저장 완료: {filename}')
```
