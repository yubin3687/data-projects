```python
import pandas as pd
from sqlalchemy import create_engine

# MySQL 연결 문자열 (본인 환경에 맞게 수정)
engine = create_engine('mysql+pymysql://root:dbqls0425!@localhost:3306/mydatabase')

# 파일명 리스트와 테이블명 리스트
file_list = [
    "C:/Users/USER/Desktop/data/cleaned/inventory_items_EDA.csv",
    "C:/Users/USER/Desktop/data/cleaned/distribution_centers_EDA.csv",
    "C:/Users/USER/Desktop/data/cleaned/events_EDA.csv",
    "C:/Users/USER/Desktop/data/cleaned/order_items_EDA.csv",
    "C:/Users/USER/Desktop/data/cleaned/orders_EDA.csv",
    "C:/Users/USER/Desktop/data/cleaned/products_EDA.csv",
    "C:/Users/USER/Desktop/data/cleaned/users_EDA.csv"
]

table_names = [
    'inventory_items',
    'distribution_centers',
    'events',
    'order_items',
    'orders',
    'products',
    'users'
]

for file, table in zip(file_list, table_names):
    df = pd.read_csv(file)
    df.to_sql(name=table, con=engine, if_exists='replace', index=False)
    print(f"{table} 테이블에 {df.shape[0]}건 저장 완료")
```
