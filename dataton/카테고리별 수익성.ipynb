```python
import pandas as pd

# 1. 데이터 불러오기
df = pd.read_csv('C:/Users/USER/Desktop/data/cleaned/inventory_items_cleaned.csv')
```

```python
# 2. 순이익과 마진율 컬럼 생성
df['profit'] = df['product_retail_price'] - df['cost']
df['margin_rate'] = df['profit'] / df['product_retail_price']
```

```python
# 3. 카테고리별 요약
category_summary = df.groupby('product_category').agg(
    avg_profit=('profit', 'mean'),
    avg_margin_rate=('margin_rate', 'mean'),
    product_count=('product_id', 'count')
).reset_index()
```

```python
# 4. 결과 저장
category_summary.to_csv('C:/Users/USER/Desktop/data/category_summary.csv', index=False)
```
