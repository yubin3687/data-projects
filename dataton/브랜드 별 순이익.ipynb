```python
# 1. 데이터 불러오기
df = pd.read_csv('C:/Users/USER/Desktop/data/cleaned/inventory_items_cleaned.csv')

# 2. 순이익 컬럼 만들기
df['profit'] = df['product_retail_price'] - df['cost']

# 3. 브랜드별 평균 순이익 계산
brand_avg_profit = df.groupby('product_brand')['profit'].mean().reset_index()

# 4. 내림차순 정렬 (이익 큰 브랜드부터)
brand_avg_profit = brand_avg_profit.sort_values(by='profit', ascending=False)

# 5. 출력
print(brand_avg_profit)
```

```python
brand_avg_profit.to_csv('C:/Users/USER/Desktop/data/brand_avg_profit.csv', index=False)
```
