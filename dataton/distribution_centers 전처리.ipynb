```python
import pandas as pd
import shutil

# 경로 설정
original_path = 'C:/Users/USER/Desktop/data/original_file/distribution_centers.csv'
copy_path = 'C:/Users/USER/Desktop/data/original_file/distribution_centers_copy.csv'

# 파일 복사
shutil.copyfile(original_path, copy_path)
print("✅ 복사 완료: distribution_centers_copy.csv")

# 1. 데이터 불러오기
df = pd.read_csv('C:/Users/USER/Desktop/data/original_file/distribution_centers_copy.csv')

# 2. 컬럼별 결측치 비율 계산
missing_ratio = df.isnull().mean()

# 3. 결측치 비율 20% 넘는 컬럼만 추출
cols_to_drop = missing_ratio[missing_ratio > 0.2].index

print("❌ 삭제할 컬럼들:", list(cols_to_drop))

# 4. 해당 컬럼 제거
df.drop(columns=cols_to_drop, inplace=True)

# 5. 전처리된 파일 저장
df.to_csv('C:/Users/USER/Desktop/data/distribution_centers_cleaned.csv', index=False)

print("✅ 전처리 완료! cleaned 파일로 저장됨.")
```
